```python
# Module 2: Missing Value Analysis & Treatment
```

```python
# Missing values before any treatment
df.isnull().sum()
```

```python
# Fill simple categorical gaps with "Unknown"
df["director"] = df["director"].fillna("Unknown")
df["director"]
```

```python
df["cast"] = df["cast"].fillna("Unknown")
df["cast"]
```

```python
df["country"] = df["country"].fillna("Unknown")
df["country"]
```

```python
# Convert date_added to datetime, then fill missing dates with Jan 1 of the mode year
df["date_added"] = pd.to_datetime(df["date_added"].str.strip(), errors="coerce")
mode_year = df["date_added"].dt.year.mode()[0]
fill_date = pd.Timestamp(year=int(mode_year), month=1, day=1)
df["date_added"] = df["date_added"].fillna(fill_date)
df["date_added"]
```

```python
# Data-quality fix: some rows have "rating" and "duration" swapped
# (rating holds values like "74 min"). Move them to the correct column first.
shifted_mask = df["rating"].astype(str).str.contains("min", na=False)
shifted_mask.sum()
```

```python
df.loc[shifted_mask, "duration"] = df.loc[shifted_mask, "rating"]
df.loc[shifted_mask, "rating"] = np.nan
```

```python
# Fill remaining missing rating values with the mode, and duration with "Unknown"
rating_mode = df["rating"].mode()[0]
df["rating"] = df["rating"].fillna(rating_mode)
df["rating"]
```

```python
df["duration"] = df["duration"].fillna("Unknown")
df["duration"]
```

```python
df.isnull().sum()
```

```python
# Module 3: Duplicate & Inconsistent Data Handling
```

```python
# Correct data types
# •	Convert date_added to datetime64
df["date_added"] = df["date_added"].astype("datetime64[ns]")

df["date_added"]
```

```python
# •	Convert release_year to int
df["release_year"] = df["release_year"].astype(int)
df["release_year"]
```

```python
# •	Convert type, rating, country to category
df["type"] = df["type"].astype("category")
df["type"]
```

```python
df["rating"] = df["rating"].astype("category")
df["rating"]
```

```python
df["country"] = df["country"].astype("category")
df["country"]
```

```python
# Save the cleaned dataset as required by the project brief
df.to_csv("netflix_cleaned.csv", index=False)
df
```

```python
# •	Countplot: Movies vs TV Shows
```

```python
plt.figure(figsize=(6, 5))
ax = sns.countplot(data=df, x="type", hue="type", palette="Set2", legend=False, dodge=False)
ax.set_title("Netflix Content: Movies vs TV Shows", fontsize=13, fontweight="bold")
ax.set_xlabel("Content Type")
ax.set_ylabel("Number of Titles")
for c in ax.containers:
    ax.bar_label(c)
plt.tight_layout()
plt.show()
```

```python
# •	Barplot: Top 15 countries with most content
```

```python
top_countries = (
    df["country"].astype(str).str.split(", ").explode().value_counts()
    .drop(labels=["Unknown"], errors="ignore")
    .head(15))

plt.figure(figsize=(9, 6))
ax = sns.barplot(x=top_countries.values, y=top_countries.index,
                  hue=top_countries.index, palette="viridis", legend=False, dodge=False)
ax.set_title("Top 15 Countries by Number of Netflix Titles", fontsize=13, fontweight="bold")
ax.set_xlabel("Number of Titles")
ax.set_ylabel("Country")
plt.tight_layout()
plt.show()
```

```python
# •	Histogram: Distribution of release_year
```

```python
plt.figure(figsize=(9, 5))
ax = sns.histplot(df["release_year"], bins=30, kde=False, color="Green")
ax.set_title("Distribution of Content by Release Year", fontsize=13, fontweight="bold")
ax.set_xlabel("Release Year")
ax.set_ylabel("Number of Titles")
plt.tight_layout()
plt.show()
```

```python
# •	Countplot: Most common rating categories
```

```python
### 5 Key Business Insights

# 1. **Movies dominate the catalog** — about 70% of titles are movies vs. 30% TV shows, so Netflix's
   # library composition still skews heavily toward film content.

```

```python
# 2. **The US is the single largest content source** — roughly 42% of titles are linked to the
   # United States (co-productions included), followed distantly by India and the UK.
```

```python
# 3. **Library growth is recent and steep** — the number of titles *added* to the platform peaked
#    in 2019-2020, showing the bulk of the catalog was built up in just the last few years of
#    available data rather than being evenly spread over time.
```

```python

# 4. **Mature-audience content leads ratings** — TV-MA is the most common rating (~37% of titles),
#    indicating the catalog leans toward adult/mature audiences over family content.

```

```python
# 5. **International content and a concentrated set of prolific creators stand out** —
#    "International Movies" is the single largest genre tag, and a small group of directors/actors
#    (e.g. Rajiv Chilaka, Anupam Kher, Shah Rukh Khan) account for a disproportionate share of titles,
#    reflecting strong regional (especially Indian) content partnerships.

```

```python
# **Conclusion:
#              ** Netflix's catalog in this dataset is movie-heavy, US-anchored but increasingly
# international, was scaled up rapidly in a short recent window, and skews toward mature-rated
# content — insights that can inform future acquisition, licensing, and regional content-investment
# strategy.
```
